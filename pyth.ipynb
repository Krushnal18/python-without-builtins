Implement binary search without using built-in methods

```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid +1
        else:
            high = mid - 1
        
    return -1

arr = [2,4,6,8,10,12,14]
print(binary_search(arr, 10))
print(binary_search(arr,5))
```

```python
def merge_sorted(list1, list2):
    merged =[]
    i = j = 0

    while i < len(list1) and j < len(list2):
        if list1[i] <= list2[j]:
            merged.append(list1[i])
            i += 1
        else:
            merged.append(list2[j])
            j += 1

        while i < len(list1):
            merged.append(list1[i])
            i += 1
        while j < len(list2):
            merged.append(list2[j])
            j += 1
        return merged

print(merge_sorted([1,3,5], [2,4,6]))
print(merge_sorted([1, 2], [3, 4, 5, 6]))
```

```python
def matrix_multiply(A, B):
    rows_A = len(A)
    cols_A = len(A[0])
    cols_B = len(B[0])
    
    # create result matrix filled with 0s
    result = [[0 for _ in range(cols_B)] for _ in range(rows_A)]
    
    for i in range(rows_A):          # row of A
        for j in range(cols_B):      # column of B
            for k in range(cols_A):  # shared dimension
                result[i][j] += A[i][k] * B[k][j]
    
    return result


A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
print(matrix_multiply(A, B))
```

```python
def find_gcd(a,b):
    while b != 0:
        a, b = b,a % b
        return a 

print(find_gcd(48,18))
print(find_gcd(100,75))
```

```python
def power(base, exponent):
    result = 1
    for _ in range(exponent):
        result *= base
    return result
print(power(2,5))
print(power(3,4))
```

```python
def decimal_to_binary(n):
    if n == 0:
        return "0"

    binary = ""
    while n > 0:
        remainder = n % 2
        binary = str(remainder) + binary
        n = n // 2

    return binary

print(decimal_to_binary(10))
print(decimal_to_binary(25))
print(decimal_to_binary(50))
```

```python
def binary_to_decimal(binary_str):
    decimal = 0
    power = len(binary_str) - 1

    for digit in binary_str:
        decimal += int(digit) * (2 ** power)
        power -= 1

    return decimal

print(binary_to_decimal("1010"))
print(binary_to_decimal("11001"))
print(binary_to_decimal("110010"))
```

```python
def is_armstrong(num):
    digits = str(num)
    num_digits = len(digits)
    total = 0
    
    temp = num
    while temp > 0:
        digit = temp % 10          
        total += digit ** num_digits
        temp = temp // 10
    
    return total == num


print(is_armstrong(153))
print(is_armstrong(123))
print(is_armstrong(9474))
```

```python
def concatenate_strings(str1, str2):
    result = []

    for char in str1:
        result.append(char)
    for char in str2:
        result.append(char)

    return "".join(result)

print(concatenate_strings("Hello", "World"))
print(concatenate_strings("Python", "Rocks"))
```

```python
def remove_duplicates(lst):
    result = []

    for item in lst:
        if item not in result:
            result.append(item)

    return result

print(remove_duplicates([1,2,2,3,4,3,5,5,8,9]))
```
